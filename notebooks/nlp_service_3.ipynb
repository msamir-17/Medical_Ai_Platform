1. The Concept: Named Entity Recognition (NER)
NER ka matlab hai text mein se "Khaas cheezein" (Entities) pehchanna.
Text: "Patient was prescribed Metformin for Type-2 Diabetes."
AI recognizes:
Metformin ⮕ CHEMICAL/DRUG
Type-2 Diabetes ⮕ DISEASE/CONDITION

```python
from transformers import pipeline

# # 1. Load the Medical NER Pipeline
# # Model: 'allenai/scibert_scivocab_uncased' is industry-standard for science/med text
# print("Loading Medical NLP Brain (SciBERT)...")
# ner_pipeline = pipeline("ner", model="allenai/scibert_scivocab_uncased", aggregation_strategy="simple")

# 1. Load a Fine-tuned Medical NER model
print("Loading Fine-tuned Medical NER Brain...")
ner_pipeline = pipeline("ner", model="d4data/biomedical-ner-all", aggregation_strategy="simple")

def extract_medical_entities(text):
    """Detects diseases, drugs, and symptoms from text."""
    entities = ner_pipeline(text)

    # Hum result ko clean format mein store karenge
    structured_results = {
        "diseases": [],
        "medications": [],
        "others": []
    }

    for ent in entities:
        # SciBERT models usually tag as LABEL_0, LABEL_1 etc. or specific tags
        # For simplicity, we just see what it finds
        word = ent['word']
        label = ent['entity_group']

        # A simple logic to categorize (In a real app, we'd use a more specific model)
        structured_results["others"].append(f"{word} ({label})")

    return structured_results
```

```python
# Real-world messy clinical note
clinical_note = "Patient presents with symptoms of Hypertension. Prescribed Amlodipine 5mg. History of Asthma."

# Let the NLP Brain read it
found_entities = extract_medical_entities(clinical_note)

print("--- The AI's Medical Understanding ---")
print(found_entities)

```
