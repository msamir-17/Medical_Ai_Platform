```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# 1. Simulate an Extracted Medical Report (Messy Text)
report_text = """
Patient Name: Sameer. Age: 30.
Report Date: 2023-10-12.
Lab Results: Glucose is 145.2 mg/dL.
Diagnosis: The patient shows signs of early-stage Hypertension.
Doctor's Note: Advised to take Amlodipine 5mg once daily and reduce salt intake.
"""

# 2. Text Splitting (Chhoti-chhoti boti banana)
# Industry mein hum text ko 500 characters ke 'chunks' mein baant te hain
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
chunks = text_splitter.split_text(report_text)

# 3. Embeddings (Text ko Numbers mein badalna)
# Hum 'all-MiniLM-L6-v2' use karenge - ye chota hai par bahut smart hai!
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# 4. Vector Store (Dukan lagana jahan se information milegi)
# FAISS ye chunks aur embeddings ko store kar lega
vector_db = FAISS.from_texts(chunks, embeddings)

print("✅ Report Indexing Complete! Your AI has 'memorized' the report.")
```
